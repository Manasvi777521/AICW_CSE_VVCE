```python
# __init__ constructor --used to initialize
class person:
    def __init__(self,name,age,school):     ##instance variables
        self.name=name
        self.age=age
        self.school=school
    def myfun(self):
        print('hello,My name is',self.name)
        print('My age is',self.age)
        print('I went to',self.school)
```

```python
p1=person('Manu',20,'vasishta school')
p1.myfun()
```

```python
class Student:
    def __init__(self,usn,name,age):
        ##instance variable
        self.usn=usn
        self.name=name
        self.age=age
    #instance method this is to access instance variable
    def show(self):
        print('USN',self.usn,'Name',self.name,'Age',self.age)
    # instance method to modify instance variable
    def update(self,usn,age):      ## er=engagemnt rate--algorthirm will use
        self.usn=usn
        self.age=age
```

```python
#create an object
stu=Student(40,'Manu',20)
stu.show()
```

```python
stu.update(12,23)
stu.show()
```
